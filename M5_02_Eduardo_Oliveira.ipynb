# Exercício 1

#### Criação

* Crie um ficheiro Excel chamado __"dados.xlsx"__ com os dados fornecidos a seguir, utilizando a biblioteca __openpyxl__:

| Nome      | Idade | Nota Matemática | Nota Português | Nota Ciências |
|-----------|-------|-----------------|----------------|---------------|
| João      | 16    | 8               | 7              | 6             |
| Maria     | 17    | 19              | 12             | 14            |
| Pedro     | 16    | 6               | 9              | 10            |
| Ana       | 18    | 15              | 20             | 17          
* Adicione uma nova coluna chamada "Avaliação" em cada disciplina (Matemática, Português e Ciências).
* A coluna Avaliação deve conter o valor "Suficiente" se a nota for igual ou superior a 10, ou "Insuficiente" se for inferior a 10.
* Crie uma nova folha chamada "Resumo", que deve incluir:
    * O número total de alunos.
    * A quantidade de notas suficientes e insuficientes por disciplina (Matemática, Português, Ciências).
* Salve o ficheiro modificado como __"dados_analise.xlsx"__.caracteres).

```python
from openpyxl import Workbook

workbook = Workbook()
folha = workbook.active
folha.title = "Dados"

#### Planilha cabecalho
cabecalhos = [
    "Nome",
    "Idade",
    "Nota Matemática",
    "Nota Português",
    "Nota Ciências"
]

for coluna, cabecalho in enumerate(cabecalhos, start=1):
    folha.cell(row=1, column=coluna).value = cabecalho

#### Alunos
dados = [
    ["João", 16, 8, 7, 6],
    ["Maria", 17, 19, 12, 14],
    ["Pedro", 16, 6, 9, 10],
    ["Ana", 18, 15, 20, 17]
]

#### Escreve os dados na folha
for linha in dados:
    folha.append(linha)

#### Adicionar coluna
folha["F1"] = "Avaliação de Matemática"
folha["G1"] = "Avaliação de Português"
folha["H1"] = "Avaliação de Ciências"

#### Marcadores
suf_matematica = insuf_matematica = 0
suf_portugues = insuf_portugues = 0
suf_ciencias = insuf_ciencias = 0

#### Avaliações
for linha in range(2, folha.max_row + 1):

    nota = folha[f"C{linha}"].value
    if nota >= 10:
        folha[f"F{linha}"] = "Suficiente"
        suf_matematica += 1
    else:
        folha[f"F{linha}"] = "Insuficiente"
        insuf_matematica += 1

    nota = folha[f"D{linha}"].value
    if nota >= 10:
        folha[f"G{linha}"] = "Suficiente"
        suf_portugues += 1
    else:
        folha[f"G{linha}"] = "Insuficiente"
        insuf_portugues += 1

    nota = folha[f"E{linha}"].value
    if nota >= 10:
        folha[f"H{linha}"] = "Suficiente"
        suf_ciencias += 1
    else:
        folha[f"H{linha}"] = "Insuficiente"
        insuf_ciencias += 1

#### Salvar primeiro arquivo
workbook.save("dados.xlsx")

##### Resumos
resumo = workbook.create_sheet("Resumo")

resumo["A1"] = "Número total de alunos"
resumo["B1"] = len(dados)

resumo["A3"] = "Disciplina"
resumo["B3"] = "Suficientes"
resumo["C3"] = "Insuficientes"

resumo.append(["Matemática", suf_matematica, insuf_matematica])
resumo.append(["Português", suf_portugues, insuf_portugues])
resumo.append(["Ciências", suf_ciencias, insuf_ciencias])

##### Salvar segundo arquivo
workbook.save("dados_analise.xlsx")

print("Arquivos criados: dados.xlsx e dados_analise.xlsx")
```

# Exercício 2

#### Criação

* Crie o ficheiro __"dados.csv"__ com os dados fornecidos, utilizando a biblioteca csv:

| Produto   | Categoria   | Preço  | Quantidade |
|-----------|-------------|--------|------------|
| Teclado   | Informática | 20.00  | 100        |
| Mouse     | Informática | 15.00  | 200        |
| Cadeira   | Móveis      | 100.00 | 50         |
| Caneca    | Utensílios  | 5.00   | 300        |
| Mesa      | Móveis      | 150.00 | 20         |
| Cadeira   | Móveis      | 250.00 | 20         |

* Ordene os dados pela coluna "Preço", do maior para o menor.
* Filtrar os dados por produtos com preço acima de 25.
* Após ordenar e filtrar, salve os resultados em um novo ficheiro CSV chamado __"dados_filtrados.csv"__.


```python
import csv

##### Criacao do arquivo
dados = [
    ["Produto", "Categoria", "Preço", "Quantidade"],
    ["Teclado", "Informática", 20.00, 100],
    ["Mouse", "Informática", 15.00, 200],
    ["Cadeira", "Móveis", 100.00, 50],
    ["Caneca", "Utensílios", 5.00, 300],
    ["Mesa", "Móveis", 150.00, 20],
    ["Cadeira", "Móveis", 250.00, 20]
]

with open("dados.csv", "w", newline="", encoding="utf-8") as ficheiro:
    escritor = csv.writer(ficheiro)
    escritor.writerows(dados)


##### Ler arquivo
with open("dados.csv", "r", newline="", encoding="utf-8") as ficheiro:
    leitor = csv.DictReader(ficheiro)
    lista = list(leitor)

##### Ordenar pelo preço
lista_ordenada = sorted(
    lista,
    key=lambda x: float(x["Preço"]),
    reverse=True
)

##### Filtrar produtos 
lista_filtrada = [
    linha for linha in lista_ordenada
    if float(linha["Preço"]) > 25
]

##### Salvar arquivo .csv
with open("dados_filtrados.csv", "w", newline="", encoding="utf-8") as ficheiro:
    campos = ["Produto", "Categoria", "Preço", "Quantidade"]

    escritor = csv.DictWriter(ficheiro, fieldnames=campos)
    escritor.writeheader()
    escritor.writerows(lista_filtrada)

print("Os arquivos 'dados.csv' e 'dados_filtrados.csv' foram criados com sucesso.")
```

# Exercício 3

#### Criação

* Crie o ficheiro Excel "vendas.xlsx" com os dados fornecidos a seguir:

| Mês        | Vendas  |
|------------|---------|
| Janeiro    | 2000    |
| Fevereiro  | 3000    |
| Março      | 2500    |
| Abril      | 4000    |
| Maio       | 3500    |
| Junho      | 4500    |

* Utilize a biblioteca __openpyxl__ para gerar um gráfico de linha que mostre a evolução das vendas ao longo dos meses.
* Adicione o gráfico à folha, posicionando-o abaixo da tabela de vendas.
* Salve o ficheiro como __"vendas_com_grafico.xlsx"__.

```python
from openpyxl import Workbook
from openpyxl.chart import LineChart, Reference

##### Criar arquivo excel
wb = Workbook()
ws = wb.active
ws.title = "Vendas"
ws.append(["Mês", "Vendas"])

##### Dados
dados = [
    ["Janeiro", 2000],
    ["Fevereiro", 3000],
    ["Março", 2500],
    ["Abril", 4000],
    ["Maio", 3500],
    ["Junho", 4500]
]

for linha in dados:
    ws.append(linha)

##### Montar estrutura inicial
wb.save("vendas.xlsx")

##### Criar gráfico

grafico = LineChart()
grafico.title = "Evolução das Vendas"
grafico.style = 2
grafico.y_axis.title = "Vendas"
grafico.x_axis.title = "Meses"

dados_grafico = Reference(ws, min_col=2, min_row=1, max_row=7)
categorias = Reference(ws, min_col=1, min_row=2, max_row=7)
grafico.add_data(dados_grafico, titles_from_data=True)
grafico.set_categories(categorias)


##### Posicionar gráfico
ws.add_chart(grafico, "A10")

##### Salvar arquivo com grafico
wb.save("vendas_com_grafico.xlsx")
print("Os arquivos 'vendas.xlsx' e 'vendas_com_grafico.xlsx' foram criados com sucesso!")
```
