# Exercício 1

#### Criação

* Apresente a data e hora atuais e mostre no formato padrão, utilizando __datetime__.
* Formate a data atual em dois formatos diferentes: "dd/mm/aaaa" e "aaaa-mm-dd HH:MM", e mostre os resultados.
* Receba (__input__) uma data de início e uma data de término para um projeto do utilizador no formato "aaaa-mm-dd HH:MM" e calcule a duração do projeto em dias, horas, minutos e segundos.
* Receba (__input__) um número de dias de prazo para uma tarefa específica do utilizador e calcule a data de vencimento a partir da data atual, mostrando a data de vencimento no formato "aaaa-mm-dd".


```python
from datetime import datetime,timedelta

agora = datetime.now()
print(agora)

formato1 = agora.strftime("%d/%m/%Y")
print("Formato 1:", formato1)
formato2 = agora.strftime("%Y-%m-%d %H:%M")
print("Formato 2:", formato2)

inicio = input("Data do início (aaaa-mm-dd HH:MM)")
fim = input("Data do término (aaaa-mm-dd HH:MM)")

data_inicio = datetime.strptime(inicio, "%Y-%m-%d %H:%M")
data_fim = datetime.strptime(fim, "%Y-%m-%d %H:%M")

duracao = data_fim-data_inicio

dias = duracao.days
segundos_restantes = duracao.seconds

horas = segundos_restantes // 3600
minutos = (segundos_restantes % 3600) // 60
segundos = segundos_restantes % 60

print(f"Duração total do projeto: {dias} dias, {horas} horas, {minutos} minutos e {segundos} segundos.")

###########

prazo = int(input("\nNúmero de dias de prazo:"))

data_atual = datetime.now()
data_vencimento = data_atual + timedelta(days=prazo)

print("Data de vencimento:",
      data_vencimento.strftime("%Y-%m-%d"))
```

# Exercício 2

#### Criação

* Crie uma classe __Evento__ que tenha os seguintes atributos:
    * titulo: título do evento.
    * data_hora_inicio: data e hora de início do evento.
    * data_hora_fim: data e hora de término do evento.
* Crie uma classe __Agenda__ que tenha o atributo:
    * Uma lista de objetos da Classe __Eventos__.
* Adicionar o método Adicionar_evento(evento) na classe __Agenda__: adiciona um evento à agenda.
* Adicionar o método listar_eventos() na classe __Agenda__: lista todos os eventos na agenda, mostrando o título, a data e hora de início, e a data e hora de término.
* Adicionar o método proximo_evento() na classe __Agenda__: mostra o próximo evento na agenda.
* Adicionar o método eventos_do_dia(data) na classe __Agenda__: lista todos os eventos para uma data específica.

#### Experimentar

* Crie um objeto da classe __Agenda__.
* Crie pelo menos __5__ objetos da classe __Evento__ e adicione à agenda.
* Liste todos os eventos.
* Mostre o próximo evento.
* Liste todos os eventos para uma data específica.

```python
from datetime import datetime

class Evento:
    def __init__(self,titulo,data_hora_inicio,data_hora_fim):
        self.titulo = titulo
        self.data_hora_inicio = data_hora_inicio
        self.data_hora_fim = data_hora_fim

    def __str__(self):
        return (f"Título: {self.titulo}\n"
                f"Início: {self.data_hora_inicio.strftime('%d/%m/%Y %H:%M')}\n"
                f"Fim: {self.data_hora_fim.strftime('%d/%m/%Y %H:%M')}")
#####

class Agenda:
    def __init__(self):
        self.eventos = []

    def adicionar_evento(self,evento):
        self.eventos.append(evento)
        self.eventos.sort(key=lambda e: e.data_hora_inicio)

    def listar_eventos(self):
        if not self.eventos:
            print("Nenhum evento cadastrado.")
            return

        print("\n**** TODOS OS EVENTOS ***")
        for evento in self.eventos:
            print(evento)
            print("-" * 40)

    def proximo_evento(self):
        agora = datetime.now()

        for evento in self.eventos:
            if evento.data_hora_inicio >= agora:
                print("\n**** PRÓXIMO EVENTO ****")
                print(evento)
                return

        print("\nNão há próximos eventos.")

    def eventos_do_dia(self,data):
        encontrados = [
            evento for evento in self.eventos
            if evento.data_hora_inicio.date() == data.date()
        ]

        print(f"\n *** EVENTOS DO DIA {data.strftime('%d/%m/%Y')} ***")

        if not encontrados:
            print("Nenhum evento encontrado.")
            return

        for evento in encontrados:
            print(evento)
            print("-" * 40)

#####

agenda = Agenda()

evento1 = Evento(
    "Reunião",
    datetime(2026, 8, 1, 9, 0),
    datetime(2026, 8, 1, 10, 30)
)

evento2 = Evento(
    "Dentista",
    datetime(2026, 8, 2, 14, 0),
    datetime(2026, 8, 2, 18, 0)
)

evento3 = Evento(
    "Blablabla",
    datetime(2026, 8, 1, 19, 0),
    datetime(2026, 8, 1, 21, 30)
)

evento4 = Evento(
    "Teste",
    datetime(2026, 8, 5, 20, 15),
    datetime(2026, 8, 6, 23, 30)
)

evento5 = Evento(
    "xxxxxxxxx",
    datetime(2026, 8, 3, 16, 0),
    datetime(2026, 8, 4, 17, 0)
)

agenda.adicionar_evento(evento1)
agenda.adicionar_evento(evento2)
agenda.adicionar_evento(evento3)
agenda.adicionar_evento(evento4)
agenda.adicionar_evento(evento5)

agenda.listar_eventos()
agenda.proximo_evento()
agenda.eventos_do_dia(datetime(2026, 8, 1))
```
